# 📝 String Data Type

A **string** is a sequence of characters enclosed in **quotes**.

You can use:
- **Single quotes** `'Hello'`
- **Double quotes** `"World"`
- **Triple quotes** for multi-line text `'''Hello\nWorld'''`

Strings are used to **store and manipulate text** — such as names, messages, or user input.

## 🧠 Key Properties:
- Strings are **immutable** (cannot be changed in place)
- Strings are **sequences**, so they support indexing and slicing
- You can **combine** (concatenate), **repeat**, and **analyze** strings with built-in functions

```python
# Creating string variables
greeting = "Hello"
name = 'Tony'
multiline = '''This is
a multi-line string.'''

print(greeting)
print(name)
print(multiline)
```

```python
# Concatenation and repetition
full_greeting = greeting + " " + name
print("Concatenated:", full_greeting)

echo = "Ha" * 3
print("Repeated:", echo)
```

```python
# Indexing and slicing
text = "Python"

print("First letter:", text[0])
print("Last letter:", text[-1])
print("First three letters:", text[:3])
print("Every other letter:", text[::2])
```

```python
# Useful string methods
message = "  hello world  "

print("Original:", message)
print("Uppercase:", message.upper())
print("Capitalized:", message.capitalize())
print("Stripped:", message.strip())
print("Replaced:", message.replace("hello", "hi"))
print("Split:", message.split())
```

```python
# Checking content
print("Is digit:", "123".isdigit())
print("Is alpha:", "abc".isalpha())
print("Is lowercase:", "hello".islower())
print("Starts with 'he':", "hello".startswith("he"))
```
