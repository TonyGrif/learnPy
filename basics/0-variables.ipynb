# 📦 Variables
A **variable** is a name that refers to a value stored in memory.  
It allows you to **store**, **reuse**, and **manipulate** data in your programs.

You can think of a variable like a **labeled box**: you put something inside and refer to it by the label (name).

## Rules for Naming Variables

- Variable names can contain **letters**, **numbers**, and **underscores**.
- They **must begin with a letter or underscore** (not a number).
- They are **case-sensitive** (`Age` and `age` are different).
- Avoid using **Python keywords** like `if`, `for`, `class`, etc.

```python
# Variable Assignment
name = "Tony"
age = 24
cups_of_coffee = 3.4
human = True

# Accessing Variables
print(name, age, cups_of_coffee, human)
```

```python
# Change the value of an existing variable
age = 31
print("Updated Age:", age)
```

```python
# Combine variables and math
year_of_birth = 2025 - age
print("Year of Birth:", year_of_birth)
```

```python
# Assign multiple variables in one line
x, y, z = 1, 2, 3
print(x, y, z)
```

```python
# All three variables get the same value
a = b = c = 100
print(a, b, c)
```

# 🧱 Python Data Types – The Basics

In Python, **data types** tell us what kind of value a variable holds.  
Each type supports specific operations and behaviors.

We'll start with the most common and essential types:

- **`int`** – Integer numbers (e.g., 3, -7, 100)
- **`float`** – Decimal numbers (e.g., 3.14, -2.0)
- **`bool`** – Boolean values: `True` or `False`
- **`str`** – Text (e.g., "hello", 'Python')

Python is **dynamically typed**, so you don't need to declare a variable's type — it's determined automatically based on the value assigned.

Check the next two notebook for more information on each type specifically

```python
print(type(name))
print(type(age))
print(type(cups_of_coffee))
print(type(human))
```
