# 📦 Tuple Data Type

A **tuple** is a collection of values, just like a list.  
However, unlike lists, **tuples are immutable** — meaning their contents **cannot be changed** after creation.

Tuples are created using **parentheses** `()`:

```python
# Creating a tuple
coordinates = (10, 20)

# Accessing elements, staring at 0
print(coordinates[0])

# Tuples are immutable — can't change or add elements!
# coordinates[0] = 15  # ❌ This would cause an error

# Check if value in tuple
print(20 in coordinates)

# Get the length of the tuple
print(len(coordinates))
```

# 🧮 Set Data Type

A **set** is an **unordered**, **mutable** collection of **unique** elements.

Sets are created using **curly braces** `{}` or the `set()` function:

```python
# Creating a set
colors = {"red", "green", "blue"}

# Adding elements
colors.add("yellow")
print(colors)

# Removing elements
colors.remove("red")
print(colors)

# Sets automatically ignore duplicates
colors.add("blue")
print(colors)

# Get length of set
print(len(colors))

# Check existence of value in set
print("green" in colors)

# Set operations
odds = {1, 3, 5}
evens = {2, 4, 6}
primes = {2, 3, 5, 7}

print(odds.union(evens))    # All elements
print(odds.intersection(primes))  # Common elements
```
