# 🗂️ Dictionary Data Type

A **dictionary** is an **unordered**, **mutable** collection of **key-value pairs**.  
It allows you to **map one value (key)** to **another (value)**.

Dictionaries are created using **curly braces** `{}` with a `key: value` syntax

## 🧠 Key Properties:
- Keys must be unique and immutable (e.g., strings, numbers, tuples)
- Values can be any data type
- Dictionaries are unordered (before Python 3.7), but insertion-ordered since Python 3.7+

```python
# Creating a dictionary
student = {
    "name": "Tony",
    "age": 24,
    "major": "Data Science"
}

# Accessing values
print("Name:", student["name"])
print("Age:", student["age"])
print("Major:", student["major"])
```

```python
# Adding a new key-value pair
student["GPA"] = 4.0

# Updating an existing value
student["age"] = 25

print("Updated student dictionary:", student)
```

```python
# Removing a key-value pair
del student["major"]

# Or using pop()
gpa = student.pop("GPA")

print("Removed GPA:", gpa)
print("Student dictionary after deletion:", student)
```

```python
# Common dictionary methods
info = {"language": "Python", "version": 3.11}

print("Keys:", info.keys())
print("Values:", info.values())
print("Items:", info.items())
print("Is 'language' a key?", "language" in info)
```

```python
# Iterating through keys and values
for key, value in student.items():
    print(f"{key}: {value}")
```

```python
# Nested dictionaries
people = {
    "alice": {"age": 30, "city": "NY"},
    "bob": {"age": 25, "city": "LA"}
}

print("Alice's age:", people["alice"]["age"])
print("Bob's city:", people["bob"]["city"])
```
