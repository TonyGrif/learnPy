# 🧱 Classes and Object-Oriented Programming (OOP)

A **class** is a blueprint for creating **objects**.  
Objects are instances of classes and can have **attributes (data)** and **methods (functions)**.

## 🧠 Why Use Classes?
- Organize code in a reusable way
- Represent real-world objects
- Bundle data and behavior together

## Terminology
- A **class** defines the structure and behavior of an object.
- **Objects** are instances of classes.
- **Attributes** are data attached to objects.
- **Methods** are functions that belong to a class.

```python
# Defining a class
class Dog:
    # The constructor
    # Contains code that runs when an object is created
    def __init__(self, name, breed):
        # self.{} are the attributes of the object
        self.name = name
        self.breed = breed

    # Class function/method
    # Any 'object' of a class can call this function
    def speak(self):
        return f"{self.name} says woof!"

# Create an object
my_dog = Dog("Buddy", "Golden Retriever")

# Use the object
print(my_dog.name)
print(my_dog.speak())
```

## 🔗 Inheritance

A class can **inherit** from another class to reuse or extend its behavior.

```python
# Create a new class inheriting all the traits of the Dog class
class Puppy(Dog):
    def __init__(self, name, breed, age):
        # Pass certain parameters to the Dog class constructor
        super().__init__(name, breed)

        self.age = age

    def is_puppy(self):
        return self.age < 2

# Create a Puppy instance
pup = Puppy("Max", "Labrador", 1)

print(pup.speak())
print("Is Max a puppy?", pup.is_puppy())
```

## 🧱 Composition

**Composition** means a class contains instances of other classes.  
It models a "has-a" relationship (e.g., a Car **has a** Engine).

```python
class Engine:
    def __init__(self, horsepower):
        self.horsepower = horsepower

    def start(self):
        return "Engine started!"

class Car:
    def __init__(self, brand, horsepower):
        self.brand = brand

        # Create a class variable that is of the type 'Engine'
        self.engine = Engine(horsepower)

    def drive(self):
        return f"{self.brand} is driving. {self.engine.start()}"

# Create a Car instance
my_car = Car("Toyota", 150)
print(my_car.drive())
```
