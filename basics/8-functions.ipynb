# 🛠️ Functions

A **function** is a reusable block of code that performs a specific task.

Functions help us:
- **Organize code** into smaller chunks
- **Reuse code** instead of rewriting it
- **Make programs easier to read and maintain**

```python
# Defining a simple function using the 'def' keyword
def greet():
    print("Hello, world!")

# Calling the function
greet()
```

```python
# Defining a function with parameters
def greet_person(name):
    print(f"Hello, {name}!")

greet_person("Tony")
```

```python
# Defining a function with a return value
def add(a, b):
    return a + b

result = add(3, 5)
print(result)
```

```python
# Defining a function with default parameter
def greet_default(name="friend"):
    print(f"Hello, {name}!")

greet_default()           # Uses default
greet_default("Charlie")  # Overrides default
```

```python
# Defining a function with multiple return values
def calculate(a, b):
    sum = a + b
    product = a * b
    return sum, product

s, p = calculate(4, 5)
print(f"Sum: {s}, Product: {p}")
```

```python
# Defining a function with a docstring
# Docstrings help people to understand what your function is doing and aids in readability
def square(n):
    """Returns the square of a number."""
    return n * n

print(square(6))
print(square.__doc__)  # Access the docstring
```

```python
# Variables are 'scoped' to the function they are created in
# This means, you can only access a variable in the function it was created in
def another_function():
    a = 100  # Local to this function

another_function()
# print(a)  # Uncommenting this line would cause an error: NameError: name 'a' is not defined
```
