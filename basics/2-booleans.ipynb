# ✅ Boolean Data Type

In Python, the `bool` data type represents **Boolean values**:  
- `True`  
- `False`

These are used to represent **truth values** and are commonly used in:
- **Conditions**
- **Comparisons**
- **Control flow**

## 🎯 Key Points:
- Booleans are capitalized: `True`, `False`
- They can be the result of a **comparison** (`==`, `>`, `<`, etc.)
- You can use **logical operators** with booleans: `and`, `or`, `not`
- Python treats `True` as `1` and `False` as `0` in numeric contexts

## Comparison Operations
These operators compare two values and return a **boolean** result (**True** or **False**).

```python
# Equal to
print(5 == 5)

# Not equal to
print(5 != 3)

# Greater than
print(7 > 2)

# Less than
print(3 < 8)

# Greater than or equal to
print(5 >= 5)

# Less than or equal to
print(4 <= 6)
```

## Logical Operations
Logical operators allow us to **combine multiple conditions** and make more complex decisions.

They always work with **Boolean values** (`True` or `False`).

```python
# Both condtions must be True
print((5 > 2) and (7 < 10))

# One condition must be True
print((5 < 2) or (7 < 10))

# Reverse the result
print(not (5 > 2))
```
