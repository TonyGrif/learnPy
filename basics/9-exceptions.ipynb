# 🚨 Exceptions

An **exception** is an error that occurs during the execution of a program.  
When an error occurs, Python stops the program and throws an **exception**.

Instead of letting the program crash, we can handle exceptions using **try-except blocks**.

## 🧠 Why Handle Exceptions?
- To prevent program crashes
- To handle invalid input or unexpected behavior
- To provide user-friendly error messages

## 🔧 Basic Syntax:
```python
try:
    # code that might raise an exception
except SomeError:
    # code that runs if the error occurs
```

See the full list of built-in exceptions [here](https://docs.python.org/3/library/exceptions.html)

````python
# Basic try-except
try:
    x = int(input("Enter a number: "))
    print("You entered:", x)
except ValueError:
    print("That's not a valid number!")
````

````python
# Handling Multiple Exceptions
try:
    a = int(input("Enter a number: "))
    b = int(input("Enter another number: "))
    result = a / b
    print("Result:", result)
except ValueError:
    print("Please enter only numbers.")
except ZeroDivisionError:
    print("You can't divide by zero.")
````

````python
# Using else and finally
try:
    num = int(input("Enter a positive number: "))
    assert num > 0
except (ValueError, AssertionError):
    print("That was not a valid positive number.")
else:
    print("Great! You entered:", num)
finally:
    print("Program finished.")
````

````python
# Raising Your Own Exceptions
def check_age(age):
    if age < 0:
        raise ValueError("Age can't be negative.")
    else:
        print("Valid age:", age)

try:
    check_age(-5)
except ValueError as e:
    print("Error:", e)
````
