# 📚 Python List (`list`) Data Type

A **list** is a collection of items in a specific order.  
It is one of Python’s most flexible built-in data structures.

You create a list using **square brackets** `[]`, and items are separated by **commas**:

## 🧠 Key Properties:
- Lists are ordered
- Lists are mutable (you can change them after creation)
- You can access, add, remove, and sort items

```python
# Creating a list
fruits = ["apple", "banana", "cherry"]
numbers = [1, 2, 3, 4, 5]

# Creating a list of different types
mixed = [42, "hello", True, 3.14]
```

```python
# Accessing elements by index
print("First number:", mixed[0])
print("Last fruit:", fruits[-1])
```

```python
# Slicing
print("First three elements:", mixed[:3])
print("Every other elements:", mixed[::2])
```

```python
# Appending and inserting
fruits.append("orange")         # Add to the end
fruits.insert(1, "blueberry")   # Insert at position 1

print(f"Updated fruits: {fruits}")
```

```python
# Removing elements
fruits.remove("banana")   # Remove by value
last = fruits.pop()       # Remove last item and return it
print(f"Removed last number: {last}")

print(f"Updated fruits: {fruits}")
```

```python
# Useful list functions
print("Length of numbers:", len(numbers))
print("Max of numbers:", max(numbers))
print("Min of numbers:", min(numbers))
print("Sum of numbers:", sum(numbers))

# Sorting
numbers.sort()
print("Sorted numbers:", numbers)

# Reversing
numbers.reverse()
print("Reversed numbers:", numbers)
```

```python
# Membership test
print("apple" in fruits)

# Looping through a list
for fruit in fruits:
    print(f"Fruit: {fruit}")
```
