# 🔄 Control Structures
Control structures allow us to **control the flow of a program** based on conditions or repetition.  
They help us make decisions or repeat certain actions.

## If Statements
If statements allow us to execute code **only if** a condition is True

### Indentation
In the following cell, notice how there are no curly braces like other languages; Python utilizes indentation to signify when a statement is over

```python
number = 5
animals = ["cat", "dog", "lizard"]

# If statement
if number == 5:
    print("Yes!")

# If/Else statement
if number > 5:
    print("Yes2!")
else:
    print("No2!")

# If/Elif/Else statement
if number < 5:
    print("Less Than!")
elif number == 5:
    print("Equal To!")
else:
    print("Greater Than!")

# Nested If
if number >= 5:
    if "dog" in animals:
        print("Both!")
    else:
        print("One!")
else:
    print("None!")
```

## For Loops
A **`for` loop** is used to **iterate** (go through) a sequence like a list, tuple, dictionary, set, or string.

It allows us to **repeat** an action for **each item** in the sequence.

```python
fruits = ["apple", "banana", "cherry"]
person = {
    "name": "Alice",
    "age": 25,
    "city": "New York"
}

# Iterating over a list
for fruit in fruits:
    print(fruit)

# Iterating with enumerate
# Enumerate gives us the index as well as the value
for index, fruit in enumerate(fruits):
    print(index, fruit)

# Iterate over a dictionary
for key, value in person.items():
    print(f"{key}: {value}")

# range(start, stop) creates numbers from start to stop-1
for i in range(5):
    print(i)

# Nested for loop
for i in range(1, 4):
    for j in range(1, 4):
        print(i * j, end=" ")
    print()
```

## While Loop
A **`while` loop** keeps running **as long as a certain condition is `True`**.

Instead of iterating over a sequence (like a `for` loop), it **depends on a condition** to continue running.

### Infinite Loop
Be careful to ensure there is an end somewhere or else the program will run forever!

```python
num = 3
values = [1, 2, 3, 4]

# While loop
while num > 1:
    print(num)
    num = num - 1

# Leave a loop early
while len(values) != 0:
    values.pop()
    print(values)
    if len(values) == 2:
        break

# Continue to skip part of a loop
i = 0
while i < 6:
    i += 1
    if i == 3:
        continue
    print(i)
```
