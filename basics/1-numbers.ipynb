# 🔢 Integers & Floats
These two types of "numeric" types and are able to perform mathematical operations.

## Math Operations
In Python, we can perform **mathematical operations** easily using operators.

These operations allow us to add, subtract, multiply, divide, and more — just like we do in regular math!

```python
# Integer variable
x = 10
y = 3

# Integer operations
print("Addition:", x + y)
print("Subtraction:", x - y)
print("Multiplication:", x * y)
print("Division (float result):", x / y)

# Divide two numbers, then remove the remainder
print("Floor Division:", x // y)

# Divide two numbers, keep only the remainder
print("Modulus:", x % y)

# Raise one number to the power of another
print("Exponentiation:", x ** y)
```

```python
# Float variable
pi = 3.14159
radius = 2.5

print("pi:", pi)
print("radius:", radius)

# Float operations
area = pi * (radius ** 2)
print("Area of a circle:", area)

# Mixed operations with int and float
result = pi + 2
print("Float + Integer:", result)
```

```python
x = 5
y = 7.5

# Converting an int to a float
print(type(x))
x_float = float(x)
print(type(x_float))

# Converting a float to an int
print(type(y))
y_int = int(y)
print(type(y_int))

# Convert a float to an int with rounding
print(y)
y = round(y)
print(y)
```
