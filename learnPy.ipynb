# 🐍 Python Overview
Python is a popular, high-level programming language known for its **simplicity**, **readability**, and **versatility**.
It is widely used in many fields such as:
- Web development
- Data science and machine learning
- Automation and scripting
- Game development
- Scientific computing

## Why Learn Python?
- **Beginner-Friendly**: Its simple syntax is great for first-time programmers.
- **Large Community**: Millions of developers worldwide contribute tutorials, libraries, and support.
- **Cross-Platform**: Python works on Windows, MacOS, and Linux.
- **Great for Prototyping**: You can build and test ideas quickly with little boilerplate.

## Key Features
- **Batteries Included**: Python comes with a vast collection of libraries and modules that can be used for various tasks.
- **Powerful Libraries**: Tools like NumPy, Pandas, Flask, Django, and TensorFlow make Python extremely powerful.
- **High-level Language** : Python abstracts away many low-level details, allowing you to focus on the logic of your program without worrying about memory management or other details.
- **Interpreted Language** : Python code is executed line-by-line, making it easier to debug and test.

```python
# Creating a dictionary
person = {
    "name": "Tony",
    "age": 24,
    "cups_of_coffee": 3.4,
    "human": True,
}

# Accessing values
print(person["name"])

# Adding a new key-value pair
person["programming_languages"] = ["Python", "Java", "Nix"]
print(person)

# Updating a value
person["age"] = 26
print(person)

# Removing a key-value pair
del person["cups_of_coffee"]
print(person)

# List the 'keys'
print(person.keys())

# List the 'values'
print(person.values())

# Check if value is in dictionary
print("name" in person)

# Get the length of the dictionary
print(len(person))
```

# 🔄 Control Structures

Control structures allow us to **control the flow of a program** based on conditions or repetition.  
They help us make decisions or repeat certain actions.

## If Statements
If statements allow us to execute code **only if** a condition is True

### Indentation
In the following cell, notice how there are no curly braces like other languages; Python utilizes indentation to signify when a statement is over

```python
number = 5
animals = ["cat", "dog", "lizard"]

# If statement
if number == 5:
    print("Yes!")

# If/Else statement
if number > 5:
    print("Yes2!")
else:
    print("No2!")

# If/Elif/Else statement
if number < 5:
    print("Less Than!")
elif number == 5:
    print("Equal To!")
else:
    print("Greater Than!")

# Nested If
if number >= 5:
    if "dog" in animals:
        print("Both!")
    else:
        print("One!")
else:
    print("None!")
```

## For Loops
A **`for` loop** is used to **iterate** (go through) a sequence like a list, tuple, dictionary, set, or string.

It allows us to **repeat** an action for **each item** in the sequence.

```python
fruits = ["apple", "banana", "cherry"]
person = {
    "name": "Alice",
    "age": 25,
    "city": "New York"
}

# Iterating over a list
for fruit in fruits:
    print(fruit)

# Iterating with enumerate
# Enumerate gives us the index as well as the value
for index, fruit in enumerate(fruits):
    print(index, fruit)

# Iterate over a dictionary
for key, value in person.items():
    print(f"{key}: {value}")

# range(start, stop) creates numbers from start to stop-1
for i in range(5):
    print(i)

# Nested for loop
for i in range(1, 4):
    for j in range(1, 4):
        print(i * j, end=" ")
    print()
```

## While Loop
A **`while` loop** keeps running **as long as a certain condition is `True`**.

Instead of iterating over a sequence (like a `for` loop), it **depends on a condition** to continue running.

### Infinite Loop
Be careful to ensure there is an end somewhere or else the program will run forever!

```python
num = 3
values = [1, 2, 3, 4]

# While loop
while num > 1:
    print(num)
    num = num - 1

# Leave a loop early
while len(values) != 0:
    values.pop()
    print(values)
    if len(values) == 2:
        break

# Continue to skip part of a loop
i = 0
while i < 6:
    i += 1
    if i == 3:
        continue
    print(i)
```

# 🛠️ Functions

A **function** is a reusable block of code that performs a specific task.

Functions help us:
- **Organize code** into smaller chunks
- **Reuse code** instead of rewriting it
- **Make programs easier to read and maintain**

```python
# Defining a simple function using the 'def' keyword
def greet():
    print("Hello, world!")

# Calling the function
greet()
```

```python
# Defining a function with parameters
def greet_person(name):
    print(f"Hello, {name}!")

greet_person("Tony")
```

```python
# Defining a function with a return value
def add(a, b):
    return a + b

result = add(3, 5)
print(result)
```

```python
# Defining a function with default parameter
def greet(name="friend"):
    print(f"Hello, {name}!")

greet()           # Uses default
greet("Charlie")  # Overrides default
```

```python
# Defining a function with multiple return values
def calculate(a, b):
    sum = a + b
    product = a * b
    return sum, product

s, p = calculate(4, 5)
print(f"Sum: {s}, Product: {p}")
```

```python
# Defining a function with a docstring
# Docstrings help people to understand what your function is doing and aids in readability
def square(n):
    """Returns the square of a number."""
    return n * n

print(square(6))
print(square.__doc__)  # Access the docstring
```

```python
# Variables are 'scoped' to the function they are created in
# This means, you can only access a variable in the function it was created in
def another_function():
    a = 100  # Local to this function

another_function()
# print(a)  # Uncommenting this line would cause an error: NameError: name 'a' is not defined
```
