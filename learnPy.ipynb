# 🧠 Basic Syntax
Let's start by learning about variables, data types, and basic operations in Python.

## Variables
A **variable** is like a container where we can store data or information. We give this container a **name**, so we can refer to it later in our program.
Try editing this code to output your information!

```python
# Variable Assignment
name = "Tony"
age = 24
cups_of_coffee = 3.4
human = True

# Accessing Variables
print(name, age, cups_of_coffee, human)
```

## Data Types
In Python, **data types** tell us what kind of data a variable holds. Every value we use in Python — whether it's a number, text, or something else — has a specific type associated with it.

Data types determine what kind of operations we can perform with the data.

### Basic Data Types
- **int**: Integer numbers (5, -10, 200)
- **float**: Decimal numbers (3.14, -0.5, 5.0)
- **str**: Text ("hello", "Python is fun!")
- **bool**: True or False values

```python
print(f"Name is a {type(name)}")
print(f"Age is a {type(age)}")
print(f"Cups is a {type(cups_of_coffee)}")
print(f"Human is a {type(human)}")
```

## Built-In Data Structures
Python provides several **built-in data structures** that make it easy to store and organize collections of data.

These include:
- **Lists**: Ordered, mutable collections of items.
- **Tuples**: Ordered, immutable collections of items.
- **Dictionaries**: Unordered collections of key-value pairs.
- **Sets**: Unordered collections of unique items.

Each structure is designed for specific use cases and offers different strengths.

| Structure     | Ordered? | Mutable? | Duplicates? | Key Feature                          |
|:-------------:|:--------:|:--------:|:-----------:|:------------------------------------:|
| List          | ✅ Yes   | ✅ Yes    | ✅ Allowed   | Dynamic size, flexible types         |
| Tuple         | ✅ Yes   | ❌ No     | ✅ Allowed   | Faster, good for fixed data          |
| Dictionary    | ❌ No    | ✅ Yes    | 🔑 Keys unique | Key-value storage                   |
| Set           | ❌ No    | ✅ Yes    | ❌ No duplicates | Fast membership tests            |

---

In the next sections, we'll look at **how to use each data structure**, with examples of common operations!

```python
# Creating a list
fruits = ["apple", "banana", "cherry"]

# Accessing individual elements
# The first element starts at 0
print(fruits[0])

# Adding element to the end of the list
fruits.append("orange")
print(fruits)

# Adding element to a specific spot in the list
fruits.insert(1, "grape")
print(fruits)

# Removing elements by value
fruits.remove("banana")
print(fruits)

# Update value in list
fruits[0] = "blackberry"

# Remove the last element in a list
fruits.pop()
print(fruits)

# Sorting alphabetically
fruits.sort()
print(fruits)

# Check if value in list
print("apple" in fruits)
print("blackberry" in fruits)

# Get the length of the list
print(len(fruits))
```

```python
# Creating a tuple
coordinates = (10, 20)

# Accessing elements, staring at 0
print(coordinates[0])

# Tuples are immutable — can't change or add elements!
# coordinates[0] = 15  # ❌ This would cause an error

# Check if value in tuple
print(20 in coordinates)

# Get the length of the tuple
print(len(coordinates))
```

```python
# Creating a dictionary
person = {
    "name": "Tony",
    "age": 24,
    "cups_of_coffee": 3.4,
    "human": True,
}

# Accessing values
print(person["name"])

# Adding a new key-value pair
person["programming_languages"] = ["Python", "Java", "Nix"]
print(person)

# Updating a value
person["age"] = 26
print(person)

# Removing a key-value pair
del person["cups_of_coffee"]
print(person)

# List the 'keys'
print(person.keys())

# List the 'values'
print(person.values())

# Check if value is in dictionary
print("name" in person)

# Get the length of the dictionary
print(len(person))
```

```python
# Creating a set
colors = {"red", "green", "blue"}

# Adding elements
colors.add("yellow")
print(colors)

# Removing elements
colors.remove("red")
print(colors)

# Sets automatically ignore duplicates
colors.add("blue")
print(colors)

# Get length of set
print(len(colors))

# Check existence of value in set
print("green" in colors)

# Set operations
odds = {1, 3, 5}
evens = {2, 4, 6}
primes = {2, 3, 5, 7}

print(odds.union(evens))    # All elements
print(odds.intersection(primes))  # Common elements
```

## Optional Typing
Python is a **dynamically typed** language, meaning you do not have to specify the type when creating a variable; however, for readability and clarity, it is preferable to add **type hints** to your code.

```python
# Variable Assignment
name: str = "Tony"
age: int = 24
cups_of_coffee: float = 3.4
human: bool = True

# Accessing Variables
print(name, age, cups_of_coffee, human)

print(f"Name is a {type(name)}")
print(f"Age is a {type(age)}")
print(f"Cups is a {type(cups_of_coffee)}")
print(f"Human is a {type(human)}")
```

## Math Operations
In Python, we can perform **mathematical operations** easily using operators.

These operations allow us to add, subtract, multiply, divide, and more — just like we do in regular math!

| Operation        | Symbol | Example    | Result |
|:----------------:|:------:|:----------:|:------:|
| Addition          | `+`    | `2 + 3`    | `5`    |
| Subtraction       | `-`    | `7 - 4`    | `3`    |
| Multiplication    | `*`    | `5 * 6`    | `30`   |
| Division          | `/`    | `8 / 2`    | `4.0`  |
| Floor Division    | `//`   | `7 // 2`   | `3`    |
| Modulus (Remainder) | `%`   | `7 % 2`    | `1`    |
| Exponentiation (Power) | `**` | `2 ** 3` | `8`    |

## Comparison Operations
These operators compare two values and return a **boolean** result (**True** or **False**).

| Operator | Meaning                | Example   | Result   |
|:--------:|:----------------------:|:---------:|:--------:|
| `==`     | Equal to                | `5 == 5`  | `True`   |
| `!=`     | Not equal to            | `5 != 3`  | `True`   |
| `>`      | Greater than            | `7 > 2`   | `True`   |
| `<`      | Less than               | `3 < 8`   | `True`   |
| `>=`     | Greater than or equal to| `5 >= 5`  | `True`   |
| `<=`     | Less than or equal to   | `4 <= 6`  | `True`   |

## Logical Operations
Logical operators allow us to **combine multiple conditions** and make more complex decisions.

They always work with **Boolean values** (`True` or `False`).

| Operator | Meaning                        | Example              | Result  |
|:--------:|:-------------------------------:|:--------------------:|:-------:|
| `and`    | Both conditions must be True     | `(5 > 2) and (7 < 10)` | `True`  |
| `or`     | At least one condition is True   | `(5 < 2) or (7 < 10)`  | `True`  |
| `not`    | Reverses the result (True → False)| `not (5 > 2)`          | `False` |
